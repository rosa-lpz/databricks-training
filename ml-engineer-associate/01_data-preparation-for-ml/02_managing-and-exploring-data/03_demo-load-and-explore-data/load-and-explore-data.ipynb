````python
# Step 1: Convert Spark DataFrame to pandas DataFrame
telco_pdf = telco_df.toPandas()

# Display basic info
print(f"Shape: {telco_pdf.shape}")
print(f"\nColumn types:\n{telco_pdf.dtypes}")
display(telco_pdf.head(10))
````

````python
# Step 2a: Distribution of the target variable (Churn)
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate counts and percentages
churn_counts = telco_pdf['Churn'].value_counts()
churn_pct = telco_pdf['Churn'].value_counts(normalize=True) * 100

print("=" * 40)
print("Churn Distribution (Class Balance)")
print("=" * 40)
for label in churn_counts.index:
    print(f"  {label}: {churn_counts[label]:>5} ({churn_pct[label]:.1f}%)")
print(f"  Total: {len(telco_pdf):>5}")
print("=" * 40)

# Visualization
fig, ax = plt.subplots(figsize=(8, 5))
sns.countplot(data=telco_pdf, x='Churn', palette=['#2ecc71', '#e74c3c'], ax=ax)
ax.set_title('Distribution of Target Variable: Churn', fontsize=14, fontweight='bold')
ax.set_xlabel('Churn', fontsize=12)
ax.set_ylabel('Count', fontsize=12)

# Add count and percentage labels on bars
for p in ax.patches:
    height = p.get_height()
    ax.text(p.get_x() + p.get_width() / 2., height + 50,
            f'{int(height)}\n({height / len(telco_pdf) * 100:.1f}%)',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
````

````python
# Step 2b: Missing value percentage across columns
missing_count = telco_pdf.isnull().sum()
missing_pct = (missing_count / len(telco_pdf) * 100).sort_values(ascending=False)
missing_df = missing_pct[missing_pct > 0].reset_index()
missing_df.columns = ['Column', 'Missing_Percentage']

print("=" * 50)
print("Missing Value Analysis")
print("=" * 50)
if len(missing_df) > 0:
    print(missing_df.to_string(index=False))
else:
    print("No missing values found in any column.")
print("=" * 50)

# Visualization (only if there are missing values)
if len(missing_df) > 0:
    fig, ax = plt.subplots(figsize=(10, 6))
    sns.barplot(data=missing_df, x='Missing_Percentage', y='Column', palette='YlOrRd', ax=ax)
    ax.set_title('Missing Value Percentage by Column', fontsize=14, fontweight='bold')
    ax.set_xlabel('Missing Percentage (%)', fontsize=12)
    ax.set_ylabel('Column', fontsize=12)

    for p in ax.patches:
        ax.text(p.get_width() + 0.1, p.get_y() + p.get_height() / 2,
                f'{p.get_width():.2f}%', ha='left', va='center', fontsize=10)

    plt.tight_layout()
    plt.show()
````

````python
# Step 2c: Relationship between key numeric features and Churn
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, col in enumerate(numeric_cols):
    sns.boxplot(data=telco_pdf, x='Churn', y=col,
                palette=['#2ecc71', '#e74c3c'], ax=axes[i])
    axes[i].set_title(f'{col} vs Churn', fontsize=13, fontweight='bold')
    axes[i].set_xlabel('Churn', fontsize=11)
    axes[i].set_ylabel(col, fontsize=11)

plt.suptitle('Numeric Features Distribution by Churn', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# Summary statistics by Churn
print("\nSummary Statistics by Churn:")
print(telco_pdf.groupby('Churn')[numeric_cols].describe().T)
````

````python
# Step 2d: Distribution of important categorical features grouped by Churn
# Note: The prompt mentions 'InternalService', but the actual column is 'internetService'
categorical_cols = ['Contract', 'PaymentMethod', 'internetService']

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

for i, col in enumerate(categorical_cols):
    sns.countplot(data=telco_pdf, x=col, hue='Churn',
                  palette=['#2ecc71', '#e74c3c'], ax=axes[i])
    axes[i].set_title(f'{col} by Churn', fontsize=13, fontweight='bold')
    axes[i].set_xlabel(col, fontsize=11)
    axes[i].set_ylabel('Count', fontsize=11)
    axes[i].tick_params(axis='x', rotation=30)
    axes[i].legend(title='Churn')

plt.suptitle('Categorical Features Distribution by Churn', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# Churn rate by category
for col in categorical_cols:
    churn_rate = telco_pdf.groupby(col)['Churn'].apply(
        lambda x: (x == 'Yes').sum() / len(x) * 100
    ).sort_values(ascending=False)
    print(f"\nChurn Rate by {col}:")
    print(churn_rate)
````

### EDA Summary

**1. Is the dataset balanced or imbalanced?**

The dataset is **imbalanced**. The target variable `Churn` has 5,174 non-churned customers (73.5%) vs 1,869 churned customers (26.5%), a roughly 3:1 ratio. This imbalance should be addressed before modeling using techniques such as class weighting, SMOTE, or stratified sampling.

**2. Which features appear to influence churn the most?**

| Feature | Churned vs Non-Churned | Signal Strength |
| --- | --- | --- |
| `tenure` | Mean 18.7 vs 37.5 (lower tenure = higher churn) | Strong |
| `MonthlyCharges` | Mean 74.4 vs 61.4 (higher charges = higher churn) | Strong |
| `TotalCharges` | Mean 1,532 vs 2,555 (correlated with lower tenure) | Moderate |
| `Contract` | Month-to-month 42.7% churn vs Two year 2.8% | Very Strong |
| `PaymentMethod` | Electronic check 45.3% vs Credit card 15.2% | Strong |
| `internetService` | Fiber optic 41.5% vs No internet 7.6% | Strong |

**3. What data quality issues should be addressed before modeling?**

* **High missing value rates**: `tenure` (35.5%), `StreamingTV` (21.3%), `MonthlyCharges` (21.3%), `Partner` (14.2%), `internetService` (14.2%), `gender` (10.6%) — these require a deliberate imputation strategy (median/mode imputation, or model-based imputation for columns with high missingness)
* **Low missing rate**: `TotalCharges` (0.16%) — simple median imputation is sufficient
* **Class imbalance**: 73.5% No vs 26.5% Yes — apply resampling or class weighting to avoid biased model predictions
